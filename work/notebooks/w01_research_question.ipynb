## 2. The question: decision, action, cost of a wrong call

*What decision does your work improve? Who acts on it? What does a wrong recommendation cost?*

Decision this improves: which pages should an SEO or content editor prioritize for reviewing and improving titles, meta descriptions, and search snippets among thousands of indexed pages?

Who acts, and what they do: an SEO or content editor can use a prioritized list of pages, starting with the highest-ranked opportunities and reviewing their titles, meta descriptions, and other on-page elements before moving to lower-priority pages.

Cost of a wrong recommendation: a wrong ranking can lead to two types of problems. A false positive may cause the editor to spend time reviewing a page that does not have a meaningful CTR opportunity. A false negative can be more difficult because a page with strong visibility but unexpectedly low CTR might be overlooked. Since the editor has limited time, the system should focus on placing the most promising pages near the top of the list rather than aiming for perfect accuracy across every page.

Why data/ML helps: a simple rule like “flag pages with CTR below 1%” is not enough because CTR depends on search position. Pages in higher positions usually get more clicks than pages ranked lower.

```python
import pandas as pd

data = pd.read_csv("data/raw/content_refresh_anonymized.csv")
print("Number of pages:", len(data))
filtered_data = data[data["impressions_90d"] >= 100].copy()
percentage = len(filtered_data) / len(data)
print(
    f"Pages with 100+ impressions: {len(filtered_data)} "
    f"({percentage:.0%} of all pages)"
)

position_order = ["top_3", "page_1", "striking", "page_3_5", "deep"]
ctr_by_position = (
    filtered_data.groupby("position_tier")["ctr"]
    .agg(["median", "size"])
    .reindex(position_order)
)
print("\nCTR by position tier:")
print(ctr_by_position)
first_page = filtered_data[
    filtered_data["position_tier"] == "page_1"
]

median_ctr = first_page["ctr"].median()
lower_limit = first_page["ctr"].quantile(0.25)
potential_pages = first_page[
    first_page["ctr"] <= lower_limit
]
print(f"\nMedian CTR for page-1 results: {median_ctr:.2f}%")
print(f"Lower 25% CTR threshold: {lower_limit:.2f}%")
print(
    f"Potential review pages: {len(potential_pages)}"
)
```
