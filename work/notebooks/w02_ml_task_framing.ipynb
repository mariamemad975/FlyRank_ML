## 2. Target or proxy

*What would you predict? Where does that label come from — observed outcome or a defined rule?*

Proxy target: CTR gap = actual_ctr − expected_ctr_for_position_tier

This is the core signal for CTR/Engagement Opportunity Scoring. Expected CTR is defined as the median CTR for pages in the same position tier, calculated in previous task. It is a proxy rather than an observed outcome, because we do not have actual CTR data after improvements. The metric represents the engagement opportunity by measuring how much a page’s actual CTR falls below that of similar-position pages.

## 4. The unit of analysis, as a real dataframe

*Load your lane's slice and show it: one row = one what?*

```python
!git clone https://github.com/mariamemad975/FlyRank_ML.git repo
%cd repo
```

```python
import pandas as pd

df = pd.read_csv("data/raw/content_refresh_anonymized.csv")

scored_pages = df[df["impressions_90d"] >= 100].copy()

scored_pages["expected_ctr"] = (
    scored_pages.groupby("position_tier")["ctr"].transform("median"))

scored_pages["ctr_gap"] = (
    scored_pages["ctr"] - scored_pages["expected_ctr"])

unit_df = scored_pages[
    ["position_tier", "impressions_90d", "ctr", "expected_ctr", "ctr_gap"]]

print("One row = one indexed page, scored for CTR/Engagement Opportunity")
unit_df.sort_values("ctr_gap").head(10)
```

## 5. Why ML beats a fixed rule here

*What makes the pattern too messy for an if-statement?*

CTR/Engagement Opportunity Scoring cannot rely on a fixed rule such as “flag pages with CTR below 1%” because CTR naturally varies by search position. For example, the median CTR is 0.19% for the top_3 tier but 0.00% for deeper positions. A single threshold would over-flag pages that naturally have low CTR and miss genuine underperformers in higher positions.
As additional signals are introduced, such as device, query intent, and historical trends, a rule-based approach would become increasingly difficult to manage. A scoring model can combine multiple signals into a single, smooth, and defensible opportunity ranking. This is the core premise of CTR/Engagement Opportunity Scoring as a research lane.
